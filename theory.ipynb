## Функции ввода и вывода

Перед тем как начать писать программу, мы должны четко понимать, что она должна делать. В больших проектах — это UX/UI дизайн, ориентир на пользователя или бизнес. Но все начинается с малого. И тем не менее принцип тот же. В небольших программах (их еще называют скрипты) мы решаем определенную задачу. Алгоритм решения задачи такой же как в математике или физике — дано -> решение -> ответ. В качестве «дано» у нас выступают исходные данные, решением будет сам скрипт, а ответ — то, что программа должна вывести. Задачи могут быть разными, основными остаются этапы ввода и вывода.

В питоне ввод с клавиатуры осуществляется при помощи функции `input()`. Она принимает потоковый ввод и возвращает строку.


````python
input()
````

Также в `input()` можно передать аргумент — промпт — то, что будет показано в консоли перед тем, как ввести в нее данные. Его еще называют «приглашением ввода».


````python
input("Введите что-нибудь:")
````

А за вывод в консоль отвечает функция `print()`. Ей можно передавать любой тип данных (о них мы поговорим позже).


````python
print("Hello, World!")
````

`print()` тоже принимает аргументы. Во-первых, ей можно передать несколько объектов, которые нужно напечатать. Во вторых, мы можем управлять тем, _как_ эти объекты будут напечатаны. За это отвечают два аргумента — `sep` и `end`.


`sep` — это разделитель _между_ объектами, если мы передаем несколько. У него есть значение по умолчанию — пробел. То есть если `sep` не указать, то объекты выведутся через пробел:


````python
print("Передаем", "несколько", "объектов")            # Напечатаются через пробел
print("Передаем", "несколько", "объектов", sep="->")  # Между объектами будет стрелочка
````

`end` — то, что будет напечатано _после_ всех объектов (или одного объекта). По умолчанию, это символ новой строки, то есть несколько идущих подряд `print()` напечатают свои объекты каждый на новой строке. Но если мы хотим, чтобы ставилось, например, две или четыре строки, нужно изменить параметр `end`:


````python
print("Я выше тебя!")
print("Всего-то на одну строчку...")

print("\n---------------------------\n")

print("А теперь что скажешь?)))", end="\n\n\n\n")
print("Эх(")
````

## Какие существуют типы данных

Отлично, как отдавать и принимать данные разобрались. Теперь неплохо бы понимать, а с чем питон работает, что он вообще умеет. Языки программирования строятся на работе с объектами различных типов: числа, строки, булевый и так далее.

> Типы данных сами по себе являются просто встроенными классами. Хотя мы будем изучать их позже, понимать это лучше уже сейчас. Изучив классы, мы сможем запросто создавать собственные типы данных.

У каждого типа есть определенные свойства. Некоторые из этих свойств общие для нескольких типов, некоторые присущи только одному. В питоне большое множество различных типов, а свойств и того больше, мы рассмотрим только основные. Но как понять, к какому типу принадлежит объект? Для этого существует функция `type()`. Если совсем по-честному, она принимает объект и возвращает класс, к которому он принадлежит. Мы теперь знаем, что типы данных тоже являются классами, значит функция `type()` нам подходит.


### Числа

Первое, что приходит в голову при виде слова «программирование» — «многачисел». В самом деле, работа с числами в программировании занимает существенную часть времени. И неважно, проводим мы анализ данных или пишем сайт — работа с числами есть везде. Так как с этим обстоят дела в питоне?


#### Целые числа

Самыми простыми числами являются целые. Так что же питон может с ними делать? Мы можем:


````python
print(12 + 34)  # Складывать
print(12 - 34)  # Вычитать
print(12 * 34)  # Умножать
                # - Делить?
                # - Неа)
print(12**34)   # Возводить в степень
````

Почему ж мы так обделили деление? На самом деле деление действительно есть, но деление возвращает вещественное число, даже если оба начальных числа целые и делятся друг на друга нацело, а это уже другой тип данных. Его рассмотрим далее. Тем не менее все же есть операция, которая будет возвращать целое число при делении — целочисленное деление:


````python
print(5 // 2)  # Делить целочисленно
````

`2` полностью содержится в `5` ровно два раза. Остаток отбрасывается. Чтобы его получить, есть еще одна операция — остаток от деления:


````python
print(5 % 2)  # Получить остаток от деления
````

А теперь посмотрим на вывод функции `type()` для целых чисел:


````python
print(type(5))
````

#### Вещественные числа

Эти числа еще называют числами с плавающей точкой. Они поддерживают те же операции, что и целые числа. Вернемся к обделенным и неразделенным:


````python
print(5 / 5)
````

Как мы видим, несмотря на то, что `5` делится нацело на саму себя, ответ получился вещественным числом.


Посмотрим на тип вещественных чисел:


````python
print(type(1.0))
````

#### Необычная запись

Помимо простой записи чисел, есть еще несколько ее видов, на которых стоит остановиться. Они не имеют своего типа данных, просто представляют число по-другому:


````python
print(1e6, type(1e6))              # Экспоненциальная запись
print(1_000_000, type(1_000_000))  # Просто удобная запись разрядов
print(0b100, type(0b100))          # Двоичная система счисления
print(0o100, type(0o100))          # Восьмеричная система счисления
print(0x100, type(0x100))          # Шестнадцатиричная система счисления
````

#### Приоритет операторов

Как и в математке, в питоне есть операторы с более высоким приоритетом — это действие выполнится раньше. Приоритет операторов следующий (от более высокому к более низкому):

- `()` — скобки
- `**` — возведение в степень
- `+x`, `-x`, `~x` — унарные плюс, минус и битовое отрицание
- `*`, `/`, `//`, `%` — умножение, деление, целочисленное деление, остаток от деления
- `+`, `-` — сложение и вычитание
- `<<`, `>>` — битовые сдвиги
- `&` — битовое И
- `^` — битовое исключающее ИЛИ (XOR)
- `|` — битовое ИЛИ
- `==`, `!=`, `>`, `>=`, `<`, `<=`, `is`, `is not`, `in`, `not in` — сравнение, проверка идентичности, проверка вхождения
- `not` — логическое НЕ
- `and` — логическое И
- `or` — логическое ИЛИ


### Строки

Следующим простым типом данных являются строки. Простым только в нашем понимании — строк исторически не существовало, и на заре своего зарождения строка являлась массивом символов. Так до сих пор остается в более низкоуровневых языках. И только в более высокоуровневых языках (таких как питон) строка является полноценным объектом. Со строками мы можем:


````python
print("Hello" + "World")           # Складывать или конкатенировать (обратим внимание, что пробела нет, так как в print() попадает один объект)
print("Hello" * 5)                 # Умножать на число
print("Hello"[0])                  # Получать символ по индексу
print("Hello".find("H"))           # Получать индекс по символу
print("Hello"[1:4])                # Делать срезы
print("Hello"[3:0:-1])             # С шагом тоже можно
print("He" in "Hello")             # Проверять вхождение
print("Hello".replace("lo", "p"))  # Заменять одну подстроку на другую
print(len("Hello"))                # Вывести длину
````

Посмотрим на тип данных строки:


````python
print(type("Hello"))
````

### Булевые значения

Когда мы хотим сказать, что утверждение истинно или ложно, мы используем булевые значения. В питоне это `True` для истины и `False` — для лжи. Булевые операторы используются для проверки какого-либо условия. Например, мы можем проверить, является ли одно число больше другого или содержится ли подстрока в строке. Операторы, результатом выполнения которых является булево значение, называют логическими.


````python
print(1 < 5)
print(2 <= 2)
print(6 == 6)
print(1 < 5 == 5)  # Обратим внимание на цепочку операторов
print(True and False)
print(True or False)
print(not True)
print(not False)
````

На цепочке операторов остановимся подробнее. Эта запись не так очевидна, как кажется. Мы привыкли выполнять операторы слева направо, и по идее должно быть так:

- `1 < 5` — `True`
- `True == 5` — `False`

Но правильным результатом является `True`. Почему так?

Цепочку операторов `a < b < c` следует трактовать так: `(a < b) and (b < c)`

То есть в нашем примере `1 < 5 == 5` представляется как `(1 < 5) and (5 == 5)`, что истинно.

На примере посложнее:

`1 < 5 == 5 > 6` эквивалентно `(1 < 5) and (5 == 5) and (5 > 6)`, что является ложью, так как последнее условие не выполняется.


Посмотрим на вывод функции `type()`:


````python
print(type(True))
````

## Преобразование в другие типы данных

Работать можно не только с тем типом данных, что был изначально. Мы можем преобразовать один объект в эквивалентный объект другого типа, если преобразование поддерживается. Это важно, так как не каждый тип можно преобразовать в каждый. Более того, на самом деле объекты не преобразовываются, просто создается новый, похожий по смыслу на наш. Но как это реализовано, мы рассмотрим чуть позже. Пока посмотрим, как из объекта одного типа получить объект другого. Ранее мы использовали функцию `type()`. Она выводила нам названия классов — типов. Чтобы выполнить «преобразование», нужно указать новый тип объекта, а в скобках указать преобразовываемый:


````python
print(int("202"), type(int("202")))      # str -> int
print(str(1.2), type(str(1.2)))          # float -> str
print(int(1.2), type(int(1.2)))          # float -> int
print(float("1.2"), type(float("1.2")))  # str -> float
print(int("1.2"))                        # Неправильный объект для преобразования
````

Отдельно стоит остановиться на преобразования в булевые значения. Любой объект в питоне может быть приведен к булевому значению. Более того, он имеет свой булевый эквивалент, который будет использован при проверке условий. Он определяется следующим образом:

Истинными (`True`) считаются:

- ненулевые числа
- непустые массивы, строки, кортежи и так далее

Ложными (`False`) соответственно считаются:

- число `0`
- пустые массивы, строки, кортежи, словари и другие коллекции


````python
print(bool("Hello"))  # str -> bool
print(bool(""))
print(type(bool("")), end="\n\n")

print(bool(-1))  # int -> bool
print(bool(0))
print(type(bool(0)), end="\n\n")

print(bool(["1", "2", "3"]))  # list -> bool
print(bool([]))
print(type(bool([])), end="\n\n")
````

## Работа с памятью

Мы повторили основы языка, теперь пора перейти к тому, как осуществляется хранение создаваемых объектов. Как реализовывается динамическая типизация, что такое переменные и ссылки, — рассмотрим далее.

Когда мы используем какой-то объект, ему выделяется ячейка в оперативной памяти. У каждой ячейки есть свой адрес. Чтобы его посмотреть, существует функция `id()`:


````python
id("hello")
````

Примечательно, что даже два одинаковых объекта будут иметь разные ячейки в памяти и, как следствие, разные адреса:


````python
print(id(1000))
print(id(1000))
````

Тем не менее...


````python
print(id(67))
print(id(67))
````

Было проведено исследование, что числа от `-5` до `256` используются чаще всего, поэтому за каждым числом из этого диапазона закреплена своя ячейка памяти, и она не будет меняться в течение всей жизни программы.

Хорошо, теперь мы знаем, что питон использует такую систему хранения, чтобы оптимизировать используемые ресурсы, но каким образом он это делает? Так вот, питон может создать какой-то объект, а потом просто ссылаться на него. Это значит, что в ячейке памяти лежит только один объект, а питон может получить к нему доступ по `id` сколько угодно раз, и лишней памяти израсходовано не будет. Чтобы программистам было удобно работать, а эффективность программ не падала, придумали эти ссылки на объекты именовать. Такая именованная ссылка называется переменной. Для создания именованной ссылки существует оператор присваивания — `=`:


````python
comfy_name = 256  # Выберем число из диапазона, где ссылки меняться не будут
print(id(comfy_name))
print(id(256))
````

Также мы можем захотеть присвоить одной переменной значение другой:


````python
static_var = [1, 2, 3]
static_var_new = [1, 2, 3]
print(id(static_var))
print(id(static_var_new), end="\n\n")

dynamic_var = [1, 2, 3]
dynamic_var_new = dynamic_var
print(id(dynamic_var))
print(id(dynamic_var_new))
````

Что произошло в каждом случае? В первом мы создали два разных объекта в разных ячейках памяти. После чего каждой ссылке мы присвоили свое имя, они никак не зависят друг от друга. А во втором случае, мы создали один объект, присвоили имя ссылке на этот объект:

```
dynamic_var -> id
```

После чего мы создали еще одну ссылку, которая ссылается на первую. А первая ссылается на id:

```
dynamic_var_new -> dynamic_var -> id
```

Глобально получается следующая ситуация:

```
dynamic_var -> id
dynamic_var_new -> id
```

Ну и духота... Подумаешь! Какая разница кто куда ссылается и кто как называется? Посмотрим. Теперь изменим только переменные без суффикса `new`:


````python
static_var = [1, 2, 3]
static_var_new = [1, 2, 3]
static_var[0] = 0
print(static_var)
print(static_var_new, end="\n\n")

dynamic_var = [1, 2, 3]
dynamic_var_new = dynamic_var
dynamic_var[0] = 0
print(dynamic_var)
print(dynamic_var_new)
````

Как мы видим, изменение `static_var` не повлияло на `static_var_new`, ведь это два разных объекта, пусть и с одинаковым значением. А вот изменение `dynamic_var` отразилось и на `dynamic_var_new`, так как они ссылаются на один и тот же объект.

Чтобы проверить, совпадают ли id у переменных, существует оператор `is`. Его часто путают с оператором `==`, который сравнивает только значения:


````python
static_var = [1, 2, 3]
static_var_new = [1, 2, 3]
print(static_var == static_var_new)
print(static_var is static_var_new, end="\n\n")

dynamic_var = [1, 2, 3]
dynamic_var_new = dynamic_var
print(dynamic_var == dynamic_var_new)
print(dynamic_var is dynamic_var_new)
````

Мы научились создавать ссылки, теперь неплохо бы научиться их удалять. В этом нам поможет оператор `del`. Он удаляет саму ссылку, а объект, на который она ссылается, не трогает. Если на объект не осталось ни одной ссылки, он удаляется сборщиком мусора, но происходит это не сразу.


````python
dynamic_var = [1, 2, 3]
dynamic_var_new = dynamic_var

print("dynamic_var_new перед удалением: ", dynamic_var_new)
del dynamic_var_new
print("dynamic_var_new удалена")
print("dynamic_var остался и ссылается на прошлый объект: ", dynamic_var)
print(dynamic_var_new)  # Будет ошибка, ссылки с именем `dynamic_var_new` больше не существует
````

Еще одним примером является задача поменять переменные местами. Для этого можно использовать временную переменную для сохранения значения:


````python
a = 1
b = 2

c = a
a = b
b = c
print(a, b)
````

А можно это реализовать при помощи ссылок. В этом случае создается кортеж из двух элементов, а потом исходным переменным присваиваются ссылки на нужные объекты, надобность в третьей переменной отпадает:


````python
a = 1
b = 2
a, b = b, a
print(a, b)
````

По похожему принципу работает преобразование типов, которое мы уже рассмотрели. Когда мы его выполняем, в памяти создается новый объект — того типа, какого мы хотим, а указатель (ссылка) перемещается со старого объекта на новый.


## Изменяемые и неизменяемые типы данных

Существуют типы данных, которые мы можем как-то изменить, они называются _изменяемыми_. По той же логике типы данных, которые мы не можем менять, называются _неизменяемыми_. Где-то было что-то похожее... Вспомним, как осуществляется преобразование типов — мы просто меняем указатель, сами объекты остаются нетронутыми. Здесь смысл абсолютно тот же — изменяемыми являются объекты, которые могут управлять ссылками на другие объекты, например, списки. Список чисел — это на самом деле список указателей на числа. Создадим три численные переменные и добавим их в список:


````python
a = 1
b = 2
c = 3
number_list = [a, b, c]

print(id(a), id(number_list[0]))
print(id(b), id(number_list[1]))
print(id(c), id(number_list[2]))
````

Как мы видим, id элементов точно такой же, как у наших переменных — адреса ячеек соответствующих чисел.


Типы, которые не управляют указателями, являются неизменяемыми. К таким относятся

- целые числа
- вещественные числа
- строки
- кортежи (это последовательность указателей, но изменять ее нельзя)
- булевы значения
- байты
- `None`

Если мы попробуем изменить, например, кортеж, возникнет ошибка:


````python
mutable_array = [1, 2, 3]
mutable_array[0] = 0
print(mutable_array)  # Все хорошо, списки - изменяемый тип данных

immutable_tuple = (1, 2, 3)
immutable_tuple[0] = 0  # Ошибка: кортеж менять нельзя
````

## Аннотации типов

В питоне тип переменной указывать не обязательно — он будет вычислен автоматически во время выполения. Но такая возможность есть. Важно понимать, что это не влияет ни на корректность программы, ни на эффективность ее выполнения. Если указать переменной тип и потом присвоить объект другого типа, то может выйти предупреждение о несоответствии типов (и то это происходит не везде), но не более. Ошибки не будет, и программа продолжит свое выполнение:


````python
x: int
x = 0

y: int = "И как я тут оказалась?"

print(x, type(x))
print(y, type(y))
````

Может возникнуть вопрос, зачем вообще указывать типы данных, если они настолько бесполезны? На самом деле это совсем не так. Аннотация типов в разы повышает читаемость кода. Она поддерживается внешними инструментами (VS Code при наведении выводит окна с метаданными функции) и позволяет на раннем этапе исключать ошибки, связанные с передаче неподдерживаемого типа данных. Более того, при валидации запросов, на сервере создаются специальные модели, содержащие имена параметров и их тип данных. Если в запросе поступит не то, что ожидает сервер, то действие будет отклонено.
